```python
import trafilatura
import pandas as pd

# Target a public weather or meteorological bulletin for Ranchi
# (We are using a standard weather forecast page as our raw data source)
target_url = 'https://en.tutiempo.net/ranchi.html'

print(f"Attempting to scrape raw data from: {target_url}")

# Download the raw HTML
downloaded_html = trafilatura.fetch_url(target_url)

if downloaded_html:
    # Extract the core text, stripping away ads, navbars, and scripts
    clean_text = trafilatura.extract(downloaded_html)
    print("\n--- Successfully Extracted Text ---")
    print(clean_text[:500]) # Print the first 500 characters to verify
else:
    print("Failed to fetch the URL. Check your internet connection or the URL.")
```
